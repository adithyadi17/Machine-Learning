```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

import statsmodels.formula.api as smf
```

```python
df = sns.load_dataset("iris")
df.describe()
```

```python
df.hist(figsize=(12,10))
plt.suptitle("Feature Distributions")
plt.show()
```

```python
plt.scatter(df['sepal_length'], df['sepal_width'])
plt.xlabel("Sepal Length")
plt.ylabel("Sepal Width")
plt.title("Iris: Sepal Length vs Sepal Width")
plt.show()
```

```python
plt.boxplot(df['sepal_length'])
plt.title("Boxplot of Sepal Length")
plt.show()
```

```python
plt.figure(figsize=(10,8))
corr = df.select_dtypes(include=np.number).corr()

sns.heatmap(corr, annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```

```python
X_slr = df[['petal_length']]
y = df['sepal_length']
```

```python
X_train, X_test, y_train, y_test = train_test_split(X_slr, y, test_size=0.2, random_state=42)

print(f"X_train shape: {X_train.shape}")
print(f"X_test shape: {X_test.shape}")
print(f"y_train shape: {y_train.shape}")
print(f"y_test shape: {y_test.shape}")
```

```python
model = LinearRegression()
model.fit(X_train, y_train)

print("Linear Regression model trained successfully.")
```

```python
y_pred = model.predict(X_test)

mse = mean_squared_error(y_test, y_pred)
mae = mean_absolute_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

print(f"Mean Squared Error (MSE): {mse:.2f}")
print(f"Mean Absolute Error (MAE): {mae:.2f}")
print(f"R-squared (R2): {r2:.2f}")
```

```python
slr = LinearRegression()
slr.fit(X_train, y_train)
```

```python
y_pred_slr = slr.predict(X_test)
y_pred_slr
```

```python
print("\nSLR Results")
print("\nSLR Results")
print("R2:", r2_score(y_test, y_pred_slr))
```

```python
# Re-perform the SLR data split to ensure consistency for plotting
_, X_test_slr_plot, _, y_test_slr_plot = train_test_split(X_slr, y, test_size=0.2, random_state=42)

plt.scatter(X_test_slr_plot, y_test_slr_plot, label="Actual")
plt.plot(X_test_slr_plot, y_pred_slr, color='red', label="Predicted Line")
plt.title("SLR Fit: Petal Length vs Sepal Length")
plt.legend()
plt.show()
```

```python
X_mlr = df[['petal_length', 'sepal_width', 'petal_width']]
```

```python
X_train_mlr, X_test_mlr, y_train_mlr, y_test_mlr = train_test_split(X_mlr, y, test_size=0.2, random_state=42)

print(f"X_train_mlr shape: {X_train_mlr.shape}")
print(f"X_test_mlr shape: {X_test_mlr.shape}")
print(f"y_train_mlr shape: {y_train_mlr.shape}")
print(f"y_test_mlr shape: {y_test_mlr.shape}")
```

```python
mlr_model = LinearRegression()
mlr_model.fit(X_train_mlr, y_train_mlr)

print("Multiple Linear Regression model trained successfully.")
```

```python
y_pred_mlr = mlr_model.predict(X_test_mlr)

mse_mlr = mean_squared_error(y_test_mlr, y_pred_mlr)
mae_mlr = mean_absolute_error(y_test_mlr, y_pred_mlr)
r2_mlr = r2_score(y_test_mlr, y_pred_mlr)

print(f"Multiple Linear Regression Model Evaluation:")
print(f"  Mean Squared Error (MSE): {mse_mlr:.2f}")
print(f"  Mean Absolute Error (MAE): {mae_mlr:.2f}")
print(f"  R-squared (R2): {r2_mlr:.2f}")
```

```python
X_train, X_test, y_train, y_test = train_test_split(X_mlr, y, test_size=0.3, random_state=42)
```

```python
mlr_model_new_split = LinearRegression()
mlr_model_new_split.fit(X_train, y_train)

print("Multiple Linear Regression model trained successfully with new split.")
```

```python
y_pred_mlr_new_split = mlr_model_new_split.predict(X_test)
```

```python
print("\nMLR Results (new split)")
print("R2:", r2_score(y_test, y_pred_mlr_new_split))
```

```python
plt.figure(figsize=(10, 6))
plt.scatter(y_test, y_pred_mlr_new_split, alpha=0.7)
plt.plot([y_test.min(), y_test.max()], [y_test.min(), y_test.max()], 'r--', lw=2)
plt.xlabel('Actual Sepal Length')
plt.ylabel('Predicted Sepal Length')
plt.title('MLR Fit: Actual vs Predicted Sepal Length')
plt.grid(True)
plt.show()
```

```python
print("\nMLR Results")
print("R2:", r2_score(y_test, y_pred_mlr_new_split))
```
